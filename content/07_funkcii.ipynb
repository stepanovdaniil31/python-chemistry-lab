## Теория

### Объявление и вызов

```python
def moles(mass, molar_mass):
    return mass / molar_mass

n = moles(90, 18)
print(n)
```

**`return` vs `print`:** функция должна *возвращать* значение, а не печатать его. Тогда результат можно использовать дальше: `moles(90, 18) * 6.022e23`.

### Параметры по умолчанию и именованные аргументы

```python
def concentration(n, v_ml=1000):
    return n / (v_ml / 1000)

print(concentration(0.5))            # объём по умолчанию 1000 мл
print(concentration(0.5, 250))
print(concentration(n=0.5, v_ml=250))  # именованные аргументы — читабельнее
```
