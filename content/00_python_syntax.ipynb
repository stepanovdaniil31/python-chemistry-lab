# Python: основные конструкции


## 1. Команда, результат и комментарий

`print()` выводит значение. Текст в кавычках — строка. Комментарий начинается с `#` и поясняет код человеку.


````python
print("Привет, Python!")
print(2 + 3)
````

В одной ячейке может быть несколько команд. В этом примере они выполняются сверху вниз.


````python
# Эту строку Python не выполняет.
print("Начало")
print("Конец")  # Комментарий можно написать после команды.
````

## 2. Переменные и присваивание

Переменная — имя, связанное со значением. Знак `=` означает: вычислить правую часть и связать результат с именем слева.


````python
count = 3
sample_name = "Образец A"
print(sample_name)
print(count)
````

````python
count = 3
count = count + 1
print(count)
````

**Имена имеют значение:** `count` и `Count` — разные имена. Используем понятные имена без пробелов: `sample_name`, `temperature`. Имя не может начинаться с цифры. Имена встроенных функций вроде `print` и `list` лучше не использовать для своих переменных.


### Вычисленный результат сам не обновляется


````python
first = 10
second = 5
total = first + second
print(total)
````

````python
first = 20
print(total)
````

Чтобы получить новый результат, выражение нужно вычислить снова.


````python
total = first + second
print(total)
````

## 3. Типы значений

Тип определяет, какие операции доступны значению. Основные типы: `int` — целые числа, `float` — числа с плавающей точкой, `str` — строки, `bool` — `True` или `False`.


````python
print(type(5))
print(type(5.0))
print(type("5"))
print(type(True))
````

````python
print(5 + 2)
print("5" + "2")
````

Кавычки делают `"5"` текстом. Если нужно число, преобразуем его явно. Десятичный разделитель в коде — точка: `2.5`.


````python
text_value = "2.5"
number = float(text_value)
print(number + 1)
print(type(number))
````

Обычно объявлять тип заранее не нужно. Он принадлежит значению, с которым имя связано в данный момент.


## 4. Выражения и арифметика

Выражение вычисляется и даёт значение. Умножение записывается явно: `*`. Скобки меняют порядок действий.


````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(7 / 2)
print(2 ** 3)  # Возведение в степень.
````

`//` — деление с округлением вниз до целого; `%` — остаток. Например, семь пробирок раскладываем по три в стойку. `e` внутри числа означает умножение на степень десяти.


````python
print(7 // 3)   # Полных троек.
print(7 % 3)    # Останется пробирок.
print(1.5e3)   # 1.5 * 10**3.
print(2e-3)    # 2 * 10**(-3).
````

## 5. Строки и оформление результата

Строки хранят текст. `+` соединяет строки, `len()` сообщает длину. Текст можно заключать в одинарные или двойные кавычки.


````python
name = "Образец A"
message = "Измеряем: " + name
print(message)
print(len(name))
````

В f-строке выражения внутри `{}` подставляются в текст. Запись `:.1f` показывает число с одним знаком после десятичной точки.


````python
temperature = 23.456
print(f"Температура: {temperature:.1f} °C")
print(temperature)  # Само значение не изменилось.
````

## 6. Сравнения и логика

Сравнение даёт `True` или `False`. `==` проверяет равенство, `!=` — неравенство. Присваивание `=` и сравнение `==` выполняют разные действия.


````python
temperature = 25
print(temperature > 20)
print(temperature == 20)
print(temperature != 20)
print(20 <= temperature <= 30)
````

`and` требует выполнения обоих условий. Для `or` достаточно хотя бы одного. `not` меняет логическое значение на противоположное.


````python
is_ready = True
has_label = False
print(is_ready and has_label)
print(is_ready or has_label)
print(not has_label)
````

## 7. Ветвления: if, elif, else

`if` выполняет блок, когда условие истинно. `else` задаёт альтернативу. После условия ставится двоеточие, а строки блока выделяются одинаковым отступом — обычно четырьмя пробелами.


````python
temperature = 35
if temperature > 30:
    print("Выше заданного порога")
else:
    print("Порог не превышен")
print("Проверка закончена")
````

Последний `print` находится за пределами ветвления и выполняется при любом значении температуры.


### Несколько вариантов

`elif` проверяет следующее условие, если предыдущие оказались ложными. В одной цепочке выбирается первая подходящая ветвь.


````python
temperature = 25
if temperature < 20:
    print("Ниже диапазона")
elif temperature <= 30:
    print("В диапазоне 20–30 °C")
else:
    print("Выше диапазона")
````

### Два отдельных if

Отдельные `if` проверяются независимо. Здесь могут появиться оба сообщения.


````python
value = 12
if value > 0:
    print("Положительное число")
if value > 10:
    print("Больше десяти")
````

## 8. Списки: несколько значений под одним именем

Список записывают в квадратных скобках. Элементы разделяют запятыми. Индекс — номер позиции; отсчёт начинается с нуля.


````python
measurements = [19.8, 20.1, 20.0]
print(measurements)
print(measurements[0])
print(measurements[-1])  # Последний элемент.
print(len(measurements))
````

Список можно изменять. `append()` добавляет элемент в конец. Запись через точку означает вызов метода — действия, связанного с объектом.


````python
measurements = [19.8, 20.1, 20.0]
measurements[0] = 19.9
measurements.append(20.2)
print(measurements)
````

Для списка есть готовые операции: сумма, минимум и максимум. Срез `[:2]` берёт элементы до индекса 2, не включая его.


````python
values = [3, 8, 5]
print(sum(values))
print(min(values), max(values))
print(values[:2])
````

## 9. Цикл for: повторение для каждого элемента

`for` перебирает элементы по очереди. На каждом шаге переменная получает очередное значение, затем выполняется блок с отступом.


````python
measurements = [19.8, 20.1, 20.0]
for value in measurements:
    print(f"Измерение: {value} °C")
print("Все значения просмотрены")
````

### Заданное число повторений

`range(3)` даёт числа `0`, `1`, `2`. В `range(1, 4)` начало включено, правая граница не включена.


````python
for number in range(1, 4):
    print(f"Повтор {number}")
````

### Накопитель

Сумму можно получать по частям. `total += value` здесь означает то же, что `total = total + value`.


````python
values = [3, 8, 5]
total = 0
for value in values:
    total += value
    print("Промежуточная сумма:", total)
print("Итог:", total)
````

## 10. Цикл while: повторение по условию

`while` проверяет условие перед каждым повторением. Когда оно становится ложным, цикл заканчивается.


````python
step = 1
while step <= 3:
    print("Шаг", step)
    step += 1
print("Цикл завершён")
````

Если значение `step` не изменяется, условие остаётся истинным и цикл не завершается.


### Досрочный выход: break

`break` завершает ближайший цикл. Здесь поиск останавливается на первом значении выше порога.


````python
values = [18, 22, 35, 40]
for value in values:
    if value > 30:
        print("Первое превышение:", value)
        break
````

## 11. Словари: доступ по имени

Словарь связывает ключи со значениями. В списке мы обращались по позиции, а здесь обращаемся по ключу. Словарь записывают в фигурных скобках.


````python
sample = {"name": "A", "temperature": 23.5}
print(sample["name"])
print(sample["temperature"])
sample["temperature"] = 24.0
print(sample)
````

`in` позволяет проверить, есть ли ключ. У словаря можно перебирать пары ключ–значение с помощью `items()`.


````python
sample = {"name": "A", "temperature": 23.5}
print("temperature" in sample)
for key, value in sample.items():
    print(key, "=", value)
````

## 12. Функции: дать действию имя

`def` определяет функцию. Параметр — имя для входного значения. Код тела выполняется при вызове функции, а `return` возвращает результат вызывающему коду.


````python
def double(number):
    return number * 2
````

Определение выше сохранило функцию, но ещё не вызвало её.


````python
print(double(4))
result = double(7)
print(result + 1)
````

### Несколько параметров

При вызове подставляются аргументы — конкретные значения. Имена параметров действуют внутри функции.


````python
def mean_of_two(first, second):
    return (first + second) / 2

print(mean_of_two(20, 24))
print(mean_of_two(10, 14))
````

### print и return

`print` показывает текст. `return` передаёт значение дальше. Если функция завершилась без `return`, результатом её вызова будет `None`.


````python
def show_double(number):
    print(number * 2)

result = show_double(4)
print("Результат вызова:", result)
````

## 13. Обработка измерений

Для каждого измерения температуры проверим превышение заданного порога и посчитаем количество превышений.


### Правило проверки


````python
def above_limit(value, limit):
    return value > limit

measurements = [19.8, 20.1, 20.0, 21.3]
limit = 20.5
````

### Перебор значений и подсчёт превышений


````python
exceedances = 0
for value in measurements:
    if above_limit(value, limit):
        exceedances += 1
        print(f"{value:.1f} °C: выше порога")
    else:
        print(f"{value:.1f} °C: в пределах порога")
````

### Результат


````python
print(f"Всего измерений: {len(measurements)}")
print(f"Превышений: {exceedances}")
````

Значение, равное порогу, не считается превышением.

## 16. Ошибки

Сообщение об ошибке содержит её тип и указывает на место в коде.


### Ошибка в имени: NameError

Имена различаются даже одной буквой. Правильная переменная здесь — `value`.


````python
value = 10
# print(vlaue)
print(value)
````

### Несовместимые типы: TypeError

Строку и число нельзя складывать напрямую. Выбираем нужное преобразование.


````python
text_value = "5"
# print(text_value + 2)
print(int(text_value) + 2)
````

### Ошибка записи: SyntaxError

Если при разборе ячейки обнаружена синтаксическая ошибка, её код не начнёт выполняться.


````python
# print("Начало"
print("Начало")
````

### Обработка ожидаемой ошибки: try / except

Можно заранее описать действие на случай конкретной ошибки. Здесь программа сообщает о невозможности преобразовать текст и продолжает работу.


````python
text_value = "тепло"
try:
    temperature = float(text_value)
    print(temperature)
except ValueError:
    print("Нужно число, например 23.5")
print("Программа продолжается")
````

## 18. Основные конструкции


| Конструкция | Пример | Смысл |
|---|---|---|
| Присваивание | `count = 3` | Связать имя со значением |
| Вывод | `print(count)` | Показать значение |
| Комментарий | `# пояснение` | Оставить объяснение для человека |
| Сравнение | `count == 3` | Проверить равенство |
| Логика | `ready and has_label` | Проверить оба условия |
| Список | `[3, 8, 5]` | Хранить последовательность |
| Индекс | `values[0]` | Взять первый элемент |
| Словарь | `{"name": "A"}` | Связать ключ со значением |
| Форматирование | `f"{value:.1f}"` | Вставить число в текст |
| Преобразование | `float("2.5")` | Получить число из текста |

Блоки после `if`, `elif`, `else`, `for`, `while` и `def` начинаются с двоеточия и выделяются отступом.

```python
else:
    другое_действие
```

```python
for item in values:
    действие
```

```python

    return результат
```
