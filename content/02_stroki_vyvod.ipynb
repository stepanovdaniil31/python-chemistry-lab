### f-строки — главный инструмент сегодняшнего занятия

Внутри строки с префиксом `f` можно вставлять значения переменных в фигурных скобках:

````python
m = 14.625
print(f"Масса навески: {m} г")
print(f"Масса навески: {m:.2f} г")     # .2f — два знака после точки
print(f"Масса навески: {m:.3g} г")     # .3g — три значащие цифры
NA = 6.022e23
print(f"{NA:.3e}")                     # научная нотация
print(f"{m:>10.2f}")                   # выравнивание по ширине 10
````

Разница `.2f` и `.3g`: первый формат фиксирует знаки **после точки**, второй — **значащие цифры**. Химикам чаще нужны значащие цифры.

### Ввод с клавиатуры

````python
m = float(input("Масса, г: "))   # запустите и введите число (с ТОЧКОЙ, не запятой!)
print(f"Вы ввели {m} г")
````
