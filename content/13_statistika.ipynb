Полная обработка серии измерений — так, как её требуют в химическом практикуме: среднее, отклонение, погрешность, и строгий критерий отбрасывания выбросов (Q-тест).

## Теория

### Среднее, отклонение, погрешность

```python
import numpy as np

v = np.array([12.4, 12.6, 12.5, 12.5, 12.7])
mean = v.mean()
std = v.std(ddof=1)              # ddof=1 — выборочное отклонение (формула с n-1),
                                 # именно его требуют в лабораторном практикуме
sem = std / np.sqrt(len(v))      # стандартная погрешность среднего
print(f"V = {mean:.2f} ± {sem:.2f} мл")
```

- **Стандартное отклонение (σ)** — мера разброса серии. Большое σ = «руки дрожали».
- **Погрешность среднего** = σ/√n — чем больше измерений, тем точнее среднее.
- Результат записывают как «среднее ± погрешность».

### Q-тест: можно ли отбросить подозрительную точку?

«На глаз» — не аргумент. Формальный критерий:

1. Отсортировать серию.
2. Для подозрительной крайней точки: **Q = |подозрительная − ближайшая соседняя| / (max − min)**
3. Если Q ≥ табличного критического значения — точку отбрасывают.

Критические значения Q (доверительная вероятность 95%):

| n | 3 | 4 | 5 | 6 | 7 | 8 | 9 | 10 |
|---|---|---|---|---|---|---|---|----|
| Q крит | 0.970 | 0.829 | 0.710 | 0.625 | 0.568 | 0.526 | 0.493 | 0.466 |
