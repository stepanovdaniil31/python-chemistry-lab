### try/except — перехват ошибки

```python
try:
    m = float(input("Масса: "))    # введите что-нибудь нечисловое
    print(f"Ок: {m}")
except ValueError:
    print("Это не число!")
```

Типовой приём — «спрашивать, пока не введут нормально»:

```python
while True:
    try:
        m = float(input("Масса, г: "))
        break
    except ValueError:
        print("Повторите ввод — нужно число")
print(f"Принято: {m}")
```

### Как устроена химическая формула (с точки зрения текста)

- Элемент: заглавная буква, за которой может идти одна строчная — `H`, `He`, `Na`, `Cl`
- После элемента может стоять число атомов; если числа нет — атом один
- `H2SO4` → H×2, S×1, O×4

Полезные методы символа: `ch.isupper()`, `ch.islower()`, `ch.isdigit()`.
