## Теория

### Базовый график

```python
import numpy as np
import matplotlib.pyplot as plt

c = np.array([0.0, 0.1, 0.2, 0.3, 0.4])
A = np.array([0.02, 0.21, 0.39, 0.62, 0.80])

plt.plot(c, A, 'o')                      # 'o' — только точки
plt.xlabel("Концентрация, моль/л")
plt.ylabel("Поглощение A")
plt.title("Калибровочная кривая")
plt.grid(True)
plt.show()
```

Стили: `'o'` точки, `'-'` линия, `'o-'` и то и то, `'s--'` квадраты с пунктиром. Цвет: `color='red'` или коротко `'ro'`.

### Несколько кривых и легенда

```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), label="sin")
plt.plot(x, np.cos(x), '--', label="cos")
plt.legend()
plt.grid(True)
plt.show()
```

### Правила химического графика

- Экспериментальные точки — **маркерами**, теоретические/подобранные кривые — **линией**
- Оси всегда подписаны, с единицами измерения
- Если кривых больше одной — легенда обязательна
