## Теория

### Словарь: пары «ключ → значение»

```python
atomic = {"H": 1.008, "C": 12.011, "N": 14.007, "O": 15.999,
          "Na": 22.99, "S": 32.06, "Cl": 35.45}
print(atomic["O"])         # доступ по ключу
atomic["Fe"] = 55.85       # добавление
print("Au" in atomic)      # проверка наличия ключа
print("Fe" in atomic)
```

### Перебор словаря

```python
for element, mass in atomic.items():
    print(f"{element:>2}: {mass}")
```
