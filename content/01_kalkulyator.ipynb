## Что такое Google Colab

**Google Colab** — это среда для программирования на Python, которая работает прямо в браузере. Ничего устанавливать на компьютер не нужно: код выполняется на сервере Google, вам нужен только браузер и Google-аккаунт.

Документ, который вы сейчас читаете, называется **ноутбук** (блокнот). Он состоит из ячеек двух видов:

- **Текстовые ячейки** — объяснения, условия задач (например, эта);
- **Кодовые ячейки** — в них пишется и выполняется код на Python.

Ноутбуки хранятся на вашем Google Диске, как обычные гугл-документы.

## Первым делом: сохраните свою копию

Этот ноутбук открыт вам «только для чтения». Чтобы работать:

Откроется ваша личная копия — в ней и решайте. Сохраняется она автоматически.

2. Последнее выражение ячейки выводится и без `print` — но только последнее, поэтому привыкайте к `print`.

## Что важно знать про Colab

- **Ячейки помнят друг друга.** Переменная, созданная в одной ячейке, видна во всех, которые вы запускаете после. Поэтому запускайте ячейки **сверху вниз**.
- **Если всё сломалось/запуталось** — меню «Среда выполнения → Перезапустить и выполнить всё» (Runtime → Restart and run all). Это перезапускает всё с чистого листа, лечит 90% проблем «у меня не работает».
- **Сессия не вечна.** Если долго не работать, виртуальная машина отключается. Код и текст не пропадают (они на Диске), но переменные и созданные файлы — сбрасываются. Просто запустите ячейки заново.
- Слева есть панель со значком 📁 — там видны файлы текущей сессии (понадобится на занятии 8).

## Теория

### Python как калькулятор

Запустите ячейку и разберитесь с каждой строкой:

```python
print(2 + 3 * 4)     # приоритет операций как в математике
print(7 / 2)         # обычное деление, результат дробный
print(7 // 2)        # целочисленное деление
print(7 % 2)         # остаток от деления
print(2 ** 10)       # степень
print(6.022e23)      # научная нотация: 6.022 * 10^23
```

Научная нотация `6.022e23` вам понадобится постоянно — так записывают число Авогадро и другие очень большие и очень маленькие числа.

### Переменные

Переменная — это имя, за которым закреплено значение. Имена выбирайте осмысленные:

```python
mass = 36.0          # масса в граммах
molar_mass = 18.0    # молярная масса воды, г/моль
n = mass / molar_mass
print(n)             # количество молей
```

У каждого значения есть тип: `int` — целое, `float` — дробное. Проверить можно функцией `type()`:

```python
print(type(36))
print(type(36.0))
```

**Частая ошибка:** дробные числа пишутся с **точкой**, не с запятой. `18,0` — это для Python не число, а два числа.

### Напоминание из химии

- `n = m / M` — количество вещества (моль) = масса (г) / молярная масса (г/моль)
- `T(K) = T(°C) + 273.15`
- Число Авогадро: в одном моле 6.022×10²³ частиц
