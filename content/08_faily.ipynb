# Занятие 8. Файлы: чтение и запись

Реальные приборы — весы, спектрометры, хроматографы — выгружают результаты в файлы. Сегодня учимся их читать, обрабатывать и сохранять отчёты.

## Теория

### Запись в файл

```python
with open("data.txt", "w", encoding="utf-8") as f:
    f.write("12.4\n12.6\n12.5\n")
print("файл создан — посмотрите в панели слева (значок папки)")
```

Конструкция `with open(...) as f:` — единственный правильный способ работать с файлами: файл закроется сам.

`\n` — символ перевода строки.

### Чтение построчно

```python
with open("data.txt", encoding="utf-8") as f:
    for line in f:
        value = float(line.strip())   # strip() убирает \n и пробелы по краям
        print(value)
```

### CSV — таблица в текстовом файле

CSV — строки со значениями через разделитель (запятая или точка с запятой). Разбирается методом `split`:

```python
line = "опыт1;12.4;25.0"
parts = line.split(";")
name = parts[0]
volume = float(parts[1])
temp = float(parts[2])
print(name, volume, temp)
```

### Напоминание из химии

Титрование: к раствору с неизвестной концентрацией по каплям добавляют раствор с известной (титрант), пока индикатор не сменит цвет. Расчёт: `C_пробы = C_титранта × V_титранта / V_пробы`.

**Важно про Colab:** файлы живут, пока живёт сессия. При переподключении — пропадают, ячейки создания файлов надо перезапустить.
