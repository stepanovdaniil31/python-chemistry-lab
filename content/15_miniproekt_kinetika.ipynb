## Постановка

Вещество A превращается в B (константа скорости k₁), B превращается в C (константа k₂):

- d[A]/dt = −k₁[A] — A убывает пропорционально своему количеству
- d[B]/dt = +k₁[A] − k₂[B] — B прибывает из A и убывает в C
- d[C]/dt = +k₂[B] — C только копится

Так ведут себя радиоактивные цепочки распадов и, например, лекарство в организме: принял → всосалось в кровь → вывелось.

## Метод Эйлера

Не пугайтесь производных — численно всё просто:

**новое значение = старое значение + скорость × маленький шаг времени**

И так много-много шагов в цикле. Это и есть метод Эйлера.
