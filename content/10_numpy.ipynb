# Занятие 10. NumPy: массивы и векторные операции

NumPy — библиотека, на которой стоит вся научная обработка данных в Python. Главная идея: операция применяется **ко всему массиву сразу**, без циклов.

## Теория

### Массивы

```python
import numpy as np

masses = np.array([14.62, 14.60, 14.71, 14.58, 14.63])
moles = masses / 58.5        # ВСЕ элементы разом!
print(moles)
```

```python
print(masses.mean())    # среднее
print(masses.std())     # стандартное отклонение
print(masses.min(), masses.max())
```

### Генерация массивов

```python
print(np.arange(0, 1.01, 0.1))     # как range, но с дробным шагом
print(np.linspace(0, 100, 11))     # 11 равноотстоящих точек от 0 до 100
```

### Подбор прямой по точкам: polyfit

Метод наименьших квадратов одной строкой — Python подбирает прямую `y = k·x + b`, наилучшим образом проходящую через точки:

```python
c = np.array([0.0, 0.1, 0.2, 0.3, 0.4])       # концентрации стандартных растворов
A = np.array([0.02, 0.21, 0.39, 0.62, 0.80])  # измеренные поглощения
k, b = np.polyfit(c, A, 1)                    # 1 = степень многочлена (прямая)
print(f"A = {k:.3f}·c + {b:.3f}")
```

### Напоминание из химии

Закон Бугера—Ламберта—Бера: поглощение света раствором пропорционально концентрации, `A = ε·l·c`. На этом основана **калибровка**: измеряем A растворов с известной концентрацией, строим прямую, потом по A неизвестного раствора находим его концентрацию. Это стандартная процедура любой аналитической лаборатории.
