# Занятие 5. Списки и кортежи

В лаборатории каждое измерение повторяют несколько раз. Сегодня учимся хранить серии измерений и обрабатывать их: среднее, разброс, поиск подозрительных значений.

## Теория

### Списки

```python
masses = [14.62, 14.60, 14.71, 14.58, 14.63]
print(masses[0])       # первый элемент (нумерация с нуля!)
print(masses[-1])      # последний
print(masses[1:4])     # срез: элементы с 1 по 3
print(len(masses))     # длина
print(sum(masses))     # сумма
print(min(masses), max(masses))
```

```python
masses.append(14.65)   # добавить в конец
masses.sort()          # отсортировать НА МЕСТЕ
print(masses)
```

### Цикл по списку

```python
for m in masses:
    print(f"{m:.2f} г")
```

### Кортеж

Кортеж — «замороженный список», изменить нельзя. Удобен для пар значений:

```python
point = (25.0, 3.4)    # (температура, pH)
t, ph = point          # распаковка
print(t, ph)
```
