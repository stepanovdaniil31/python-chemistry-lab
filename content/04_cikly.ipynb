## Теория

### Цикл for и range

```python
for i in range(5):          # i пробегает 0, 1, 2, 3, 4
    print(i)
```

```python
for i in range(1, 11):      # от 1 до 10 (правая граница НЕ включается!)
    print(i, i**2)
```

### Накопитель

Классический приём — переменная, в которой копится результат:

```python
total = 0
for i in range(1, 101):
    total += i          # то же, что total = total + i
print(total)
```

### Цикл while

`for` — когда известно, сколько раз повторять. `while` — когда известно **условие остановки**:

```python
c = 1.0
steps = 0
while c > 0.001:
    c = c / 10
    steps += 1
print(f"Понадобилось {steps} разбавлений")
```

### Напоминание из химии

- Серийное разбавление: раствор разбавляют в N раз, полученный — ещё в N раз, и т.д.
- pH сильной кислоты: `pH = −log₁₀(C)`. Логарифм в Python: `import math`, затем `math.log10(x)`.
