# Занятие 3. Условия: if / elif / else

Программа начинает принимать решения: кислая среда или щелочная? лёд или пар? допустить к лабе или нет?

## Теория

### Конструкция if

```python
ph = 3.5
if ph < 7:
    print("Кислая среда")
elif ph == 7:
    print("Нейтральная")
else:
    print("Щелочная")
```

**Важно про отступы:** всё, что сдвинуто вправо после `if ...:`, выполняется только при истинном условии. Отступ — это часть синтаксиса Python, а не украшение. Стандарт — 4 пробела (Colab ставит их сам по Tab).

### Сравнения и логика

Операторы: `<` `<=` `>` `>=` `==` (равно) `!=` (не равно). Обратите внимание: проверка на равенство — **двойное** `==`.

Условия объединяются через `and`, `or`, `not`:

```python
ph = 7.2
print(0 <= ph <= 14)          # двойное неравенство — так можно!
print(ph > 6.5 and ph < 7.5)  # то же через and
```

### Напоминание из химии

- pH < 7 — кислая среда, pH = 7 — нейтральная, pH > 7 — щелочная. Шкала от 0 до 14.
- Если температура ниже точки плавления — вещество твёрдое; между плавлением и кипением — жидкость; выше кипения — газ.
