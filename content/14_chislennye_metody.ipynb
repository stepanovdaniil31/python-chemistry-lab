## Теория

### Идея бисекции

Ищем корень уравнения f(x) = 0 на отрезке [a, b], где f меняет знак:
1. Берём середину отрезка.
2. Смотрим, в какой половине происходит смена знака.
3. Оставляем эту половину и повторяем.

Каждый шаг сокращает отрезок вдвое — как угадывание числа вопросами «больше/меньше». 40 шагов уменьшают отрезок в 10¹² раз.

```python
def bisect(f, a, b, eps=1e-10):
    while b - a > eps:
        mid = (a + b) / 2
        if f(a) * f(mid) <= 0:    # смена знака в левой половине
            b = mid
        else:
            a = mid
    return (a + b) / 2

# проверка на уравнении, ответ которого знаем: x^2 = 2
def f(x):
    return x*x - 2

print(bisect(f, 0, 2))   # должно быть ~1.4142
```
