## Теория

Сначала создадим файл с данными — запустите:

```python
data = """symbol,name,mass,category,melt_c
H,Водород,1.008,неметалл,-259
C,Углерод,12.011,неметалл,3550
N,Азот,14.007,неметалл,-210
O,Кислород,15.999,неметалл,-218
Na,Натрий,22.99,металл,98
S,Сера,32.06,неметалл,115
Cl,Хлор,35.45,неметалл,-101
K,Калий,39.098,металл,64
Ca,Кальций,40.078,металл,842
Fe,Железо,55.845,металл,1538
Cu,Медь,63.546,металл,1085
Au,Золото,196.97,металл,1064"""
with open("elements.csv", "w", encoding="utf-8") as f:
    f.write(data)
print("elements.csv создан")
```

```python
import pandas as pd
df = pd.read_csv("elements.csv")
df           # в Colab таблица красиво отрисуется сама
```

### Основные операции

```python
df["mass"]                        # один столбец
```

```python
df[df["category"] == "металл"]    # фильтр по условию
```

```python
df.sort_values("mass", ascending=False)   # сортировка
```

```python
df["melt_k"] = df["melt_c"] + 273.15      # новый столбец — векторно, как numpy!
df.head()
```

```python
df.describe()    # вся статистика разом
```

**Два условия** объединяются через `&`, и каждое — в скобках: `df[(df.mass > 20) & (df.category == "металл")]`.
